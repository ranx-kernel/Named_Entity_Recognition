```python
import spacy
import pandas as pd
import gradio as gr
```

```python
nlp = spacy.load("en_core_web_sm")

print("NER model loaded successfully!")
```

```python
text = "Apple announced a new product in California on September 10, 2026."

doc = nlp(text)

print(text)
```

```python
for ent in doc.ents:
    print(ent.text, "->", ent.label_)
```

```python
for ent in doc.ents:
    print(
        "Entity:", ent.text,
        "| Label:", ent.label_,
        "| Meaning:", spacy.explain(ent.label_)
    )
```

```python
entities = []

for ent in doc.ents:
    entities.append({
        "Entity": ent.text,
        "Label": ent.label_,
        "Description": spacy.explain(ent.label_)
    })

df = pd.DataFrame(entities)

df
```

```python
text = """
Microsoft was founded in 1975.
The company is headquartered in Washington.
Satya Nadella became CEO in 2014.
"""

doc = nlp(text)

for ent in doc.ents:
    print(ent.text, "->", ent.label_)
```

```python
from spacy import displacy

displacy.render(doc, style="ent", jupyter=True)
```

```python
def extract_entities(text):
    doc = nlp(text)

    results = []

    for ent in doc.ents:
        results.append({
            "Entity": ent.text,
            "Label": ent.label_,
            "Description": spacy.explain(ent.label_) or "Unknown"
        })

    return pd.DataFrame(
        results,
        columns=["Entity", "Label", "Description"]
    )
```

```python
sample_text = """
Google was founded in 1998.
Its headquarters are located in California.
The company developed several artificial intelligence products.
"""

result = extract_entities(sample_text)

result
```

```python
with gr.Blocks() as app:

    gr.Markdown("# 🔎 Named Entity Recognition")

    gr.Markdown(
        "Enter text below to identify people, organizations, "
        "locations, dates and other named entities."
    )

    text_input = gr.Textbox(
        label="Enter Text",
        placeholder="Enter a sentence or paragraph...",
        lines=6
    )

    analyze_button = gr.Button("Extract Entities")

    output = gr.Dataframe(
        headers=["Entity", "Label", "Description"],
        datatype=["str", "str", "str"],
        label="Detected Entities"
    )

    analyze_button.click(
        fn=extract_entities,
        inputs=text_input,
        outputs=output
    )

app.launch()
```
